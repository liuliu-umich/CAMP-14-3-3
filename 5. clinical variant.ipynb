```python
# Define the regex pattern
pattern = r"\(p\.([A-Za-z]{3}\d+[A-Za-z]{3})\)"

# Filter rows based on the pattern
filtered_df = variant_summary_df[variant_summary_df["Name"].str.contains(pattern)]
filtered_df['Extracted_Pattern'] = filtered_df['Name'].str.extract(pattern)
```

```python
pattern = r"\(p\.(?P<AA1>[A-Za-z]{3})(?P<Position>\d+)(?P<AA2>[A-Za-z]{3})\)"

# Extract components and save into new columns
extracted_components = filtered_df['Name'].str.extract(pattern)

# Add the extracted columns to the original DataFrame
filtered_df['AA1'] = extracted_components['AA1']
filtered_df['Position'] = extracted_components['Position']
filtered_df['AA2'] = extracted_components['AA2']
```
