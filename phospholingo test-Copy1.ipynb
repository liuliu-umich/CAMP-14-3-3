```python
#!/usr/bin/env python
import sys
import os
from pathlib import Path

# 1. Add PhosphoLingo to Python path
phospholingo_dir = Path("src/files/PhosphoLingo/phosphoLingo").resolve()  # Update this path!
sys.path.insert(0, str(phospholingo_dir))

# 2. Import AFTER path adjustment
import predict

# 3. Define all paths relative to project root
project_root = Path.cwd()  # Assuming you're in 1433predictor/
data_dir = project_root  # Example subdirectory
print(data_dir)
# 4. File paths
model_loc = r"H:\PhosphoLingo_ST_new.ckpt"  # Raw string for Windows
dataset_fastas = [
    str(data_dir / "sequence.fasta"),
    str(data_dir / "test.fasta")
]

# 5. Create output directory
output_dir = project_root / "results"
output_dir.mkdir(exist_ok=True)

# 6. Run predictions
for dataset in dataset_fastas:
    dataset_path = Path(dataset)
    output_file = output_dir / f"{dataset_path.stem}_predictions.csv"
    
    predict.run_predict(
        model_loc,
        str(dataset_path),
        str(output_file)
    )
```
