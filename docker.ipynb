```python
import time
# Initialize Docker client
client = docker.from_env()

# List all containers
print("Listing all containers:")
for container in client.containers.list(all=True):
    print(container.name, container.status)
    
# Use the local image
image_name = "9cc1ab91e3f3"

# Run the Docker container
container = client.containers.run(
    image=image_name,
    command="ls",
    # command="python infer_1433.py",
    detach=True
)

# Wait for the container to finish executing
time.sleep(2)  # Wait for 2 seconds

# Capture and print container logs
logs = container.logs()
print(logs.decode('utf-8'))

# Stop and remove the container
container.stop()
container.remove()
```

```python
import docker

# Initialize the Docker client
client = docker.from_env()

# The name of your existing container
container_name = "plm_benji"

try:
    # Get the container by name
    container = client.containers.get(container_name)
    print(f"Starting container '{container_name}'...")

    # Start the container if it is not running
    if container.status != 'running':
        container.start()
        print(f"Container '{container_name}' started successfully.")
    else:
        print(f"Container '{container_name}' is already running.")
except docker.errors.NotFound:
    print(f"Container '{container_name}' not found.")
except docker.errors.APIError as e:
    print(f"Error starting container '{container_name}': {e}")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
    
# Run the Docker container
container = client.containers.run(
    container_name,
    command="ls",
    # command="python infer_1433.py",
    detach=True
)

# Wait for the container to finish executing
time.sleep(2)  # Wait for 2 seconds

# Capture and print container logs
logs = container.logs()
print(logs.decode('utf-8'))

# Stop and remove the container
container.stop()
container.remove()    
```
